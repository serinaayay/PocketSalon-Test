```python
import onnxruntime as rt
import numpy as np

# Load the ONNX model
b0_session = rt.InferenceSession("EfficientNetB0.onnx")
mn_session = rt.InferenceSession("MobileNetV3Small.onnx")
#rn_session = rt.InferenceSession("ResNet50.onnx") <- AYAW GUMANA >:(
```

```python
from PIL import Image

imagepath = "/content/need-product-recommendations-for-breakage-v0-xr1jh99kx7ge1.jpg"

# Load the image
img = Image.open(imagepath)

# Preprocess
img = img.resize((224, 224))
input_data = np.array(img).astype(np.float32)

# Add batch dimension
input_data = np.expand_dims(input_data, axis=0)

# Get input name from the model
input_name = b0_session.get_inputs()[0].name

# Run inference
results = b0_session.run(["hair_loss", "color_damage", "breakage"], {input_name: input_data})

# Print the output
print("Model output:", results)
```

```python
# Run inference
results = mn_session.run(["hair_loss", "color_damage", "breakage"], {input_name: input_data})

# Print the output
print("Model output:", results)
```
